```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
K = nx.MultiGraph()
pontes = [("A", "C"), ("C", "A"), ("B", "C"), ("C", "B"), ("C", "D"), ("A", "D"), ("B", "D")]
K.add_edges_from(pontes)

print("number of nodes: ", K.number_of_nodes(), "| number of edges: ", K.number_of_edges())

for v, d in sorted(K.degree()):
  print(f"grau(v) = {d} ({'odd' if d % 2 else 'even'})")

odds = sum(1 for _, d in K.degree() if d % 2)

print(f"number of odd degree nodes: {odds}")
print(f"has eulerian path: {nx.has_eulerian_path(K)}")
```

```python
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(K)
nx.draw(K, pos, with_labels=True, node_color='lightblue', edge_color='gray', node_size=1500, font_size=16, font_weight='bold')
plt.show()
```
